```python
categorical_cols=[
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course',
]

for col in categorical_cols:
  df[col]=(
      df[col]
      .astype('string')
      .str.strip()
      .str.lower()
)

for col in categorical_cols:
  print(f"\t{col}:")
  print(df[col].unique())
```

```python
score_cols=[
    'math score',
    'reading score',
    'writing score'
]
stat=df[score_cols].agg(
    ['mean','median','std']
).T
stat['Q1']=df[score_cols].quantile(0.25)
stat['Q2']=df[score_cols].quantile(0.50)
stat['Q3']=df[score_cols].quantile(0.75)
print(stat.round(2))
```

```python
df = pd.read_csv("/content/StudentsPerformance - StudentsPerformance.csv")

# Re-apply categorical cleaning to keep df consistent
categorical_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]
for col in categorical_cols:
    df[col] = df[col].astype('string').str.strip().str.lower()
```

```python
df['Total Marks'] = (
    df['math score'] + df['reading score'] + df['writing score'])
df['Percentage'] = (df['Total Marks'] / 300) * 100

# Corrected the syntax error by passing a proper list of columns to the DataFrame indexer
print(df[['math score', 'reading score', 'writing score', 'Total Marks', 'Percentage']].head())
```
